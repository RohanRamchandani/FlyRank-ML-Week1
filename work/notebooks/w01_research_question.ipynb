**Decision this improves:** which pages a content/SEO reviewer should look at first when
deciding whether to refresh, expand, protect, prune, or monitor a piece of content.

**Who acts on it:** a content or SEO team with limited review capacity — they can't manually
audit every page, so they need a ranked shortlist of the highest-priority candidates.

**Cost of a wrong call:**
- A false positive (flagging a healthy page as needing review) wastes a reviewer's limited
  time on a page that didn't need attention.
- A false negative (missing a page that is genuinely declining) means a real traffic/visibility
  problem goes unnoticed and unaddressed, potentially costing the client organic traffic and
  revenue over time.

Because review capacity is limited, the ranking itself matters as much as raw accuracy —
getting the top of the list right is more valuable than getting every single page right.
